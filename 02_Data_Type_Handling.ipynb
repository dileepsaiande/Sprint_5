```python
import pandas as pd
import numpy as np

print("Pandas version:", pd.__version__)
print("Libraries imported successfully!")
```

```python
data = {
    "Age": [25, 30, 35, 40],
    "Salary": ["25000", "35000", "invalid", "45000"],
    "Category": ["Junior", "Senior", "Junior", "Senior"],
    "Is_Employed": [True, True, False, True],
    "Joining_Date": ["2024-01-15", "2023-06-20", "invalid", "2022-03-10"],
    "Rating": [4, 5, 3, 4]
}

df_types = pd.DataFrame(data)

display(df_types)
```

```python
print("Data types of each column:")
print(df_types.dtypes)

print("\nDetailed dataset information:")
df_types.info()
```


## 1. Identifying Data Types

Data types specify the kind of values stored in a column.

- Numerical: numbers used for calculations.
- Categorical: labels or groups, such as department or city.
- Boolean: True or False values.
- Date/time: dates and timestamps.
- String: text values.

Correctly identifying data types helps prevent calculation errors
and ensures that data is processed appropriately.


```python
print("Numerical columns:")
print(df_types.select_dtypes(include=["number"]).columns.tolist())

print("\nNumerical summary:")
display(df_types.select_dtypes(include=["number"]).describe())
```

```python
print("Categorical columns:")

categorical_columns = df_types.select_dtypes(
    include=["object", "str", "category"]
).columns.tolist()

print(categorical_columns)

print("\nUnique categories:")
print(df_types["Category"].unique())
```

```python
print("Boolean column data type:")
print(df_types["Is_Employed"].dtype)

print("\nBoolean values:")
print(df_types["Is_Employed"].value_counts())

print("\nBoolean column:")
display(df_types[["Is_Employed"]])
```

```python
df_types["Is_Employed_Num"] = (
    df_types["Is_Employed"].astype(int)
)

print("Boolean converted to integer:")
display(df_types[["Is_Employed", "Is_Employed_Num"]])
```

```python
print("Current data types:")
print(df_types.dtypes)
```

## Boolean Data

Boolean data represents two logical values: True and False.

It is useful for representing conditions such as employee status,
customer activity, or whether a transaction was successful.

Boolean values can be converted into integers, where True becomes
1 and False becomes 0. This conversion should only be performed
when appropriate for the analysis or machine learning task.

```python
print("Joining_Date before conversion:")
print(df_types["Joining_Date"])

df_types["Joining_Date"] = pd.to_datetime(
    df_types["Joining_Date"],
    errors="coerce"
)

print("\nJoining_Date after conversion:")
print(df_types["Joining_Date"])

print("\nData type:")
print(df_types["Joining_Date"].dtype)
```

```python
df_types["Joining_Year"] = df_types["Joining_Date"].dt.year
df_types["Joining_Month"] = df_types["Joining_Date"].dt.month
df_types["Joining_Day"] = df_types["Joining_Date"].dt.day

display(
    df_types[
        ["Joining_Date", "Joining_Year", "Joining_Month", "Joining_Day"]
    ]
)
```


## Date/Time Data

Date/time data represents dates and timestamps.

The Pandas function `pd.to_datetime()` converts supported
date strings into datetime values.

The `errors="coerce"` argument converts invalid dates into `NaT`,
which can then be identified and handled appropriately.

Datetime values support chronological sorting and extraction
of useful features such as year, month, and day.


```python
print("Salary before conversion:")
print(df_types["Salary"])

df_types["Salary"] = pd.to_numeric(
    df_types["Salary"],
    errors="coerce"
)

print("\nSalary after conversion:")
print(df_types["Salary"])

print("\nSalary data type:")
print(df_types["Salary"].dtype)
```

```python
print("Missing values after conversion:")
print(df_types[["Salary", "Joining_Date"]].isnull().sum())

print("\nRows with missing salary:")
display(df_types[df_types["Salary"].isna()])

print("\nRows with missing joining date:")
display(df_types[df_types["Joining_Date"].isna()])
```

```python
df_types["Age"] = df_types["Age"].astype(float)

print("Age data type:", df_types["Age"].dtype)
display(df_types[["Age"]])
```


## Converting Data Types

Pandas provides several methods for converting data types.

- `astype()` converts a column to a specified data type.
- `pd.to_numeric()` converts values into numeric data.
- `pd.to_datetime()` converts supported values into datetime data.

The `errors="coerce"` argument in numeric and datetime conversion
turns invalid values into missing values instead of raising an error.

After conversion, missing values should be investigated before
deciding whether to impute them, correct them, or exclude records.


```python
print("Final dataset information:")
df_types.info()

print("\nFinal dataset:")
display(df_types)
```

```python
print("Current data types:")
print(df_types.dtypes)

print("\nCheck Salary is numeric:")
print(pd.api.types.is_numeric_dtype(df_types["Salary"]))

print("\nCheck Joining_Date is datetime:")
print(pd.api.types.is_datetime64_any_dtype(df_types["Joining_Date"]))

print("\nCheck Age is numeric:")
print(pd.api.types.is_numeric_dtype(df_types["Age"]))
```

```python
print("Missing values after data type conversions:")
display(df_types.isnull().sum())

print("\nRows with invalid or missing values:")
display(df_types[df_types.isnull().any(axis=1)])
```


## Why Correct Data Types Are Important

Correct data types help ensure that data is processed accurately
and efficiently.

1. Numerical data supports mathematical calculations.
2. Categorical data represents groups and labels.
3. Boolean data represents True or False conditions.
4. Datetime data enables chronological analysis and date-based
   feature extraction.
5. String data stores text and identifiers.

Incorrect data types can cause calculation errors, misleading
statistics, inefficient processing, and problems during machine
learning preprocessing.

Therefore, data types should be checked and validated before
training a machine learning model.
