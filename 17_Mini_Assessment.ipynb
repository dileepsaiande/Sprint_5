```python
import pandas as pd
import numpy as np

np.random.seed(42)

df = pd.DataFrame({
    "Age": np.random.randint(20, 60, 100),
    "Salary": np.random.randint(20000, 100000, 100),
    "Experience": np.random.randint(0, 25, 100),
    "Department": np.random.choice(
        ["IT", "HR", "Finance", "Marketing"], 100
    ),
    "Education": np.random.choice(
        ["Graduate", "Postgraduate", "PhD"], 100
    ),
    "Survived": np.random.choice(
        [0, 1], 100, p=[0.65, 0.35]
    )
})

# Introduce missing values.
df.loc[[2, 5, 10, 15], "Age"] = np.nan
df.loc[[3, 8, 12], "Salary"] = np.nan
df.loc[[6, 14], "Department"] = np.nan

# Introduce an extreme value.
df.loc[0, "Salary"] = 500000

print("Dataset created successfully!")
display(df.head())
```
