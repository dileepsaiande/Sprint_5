
## 1. What is Data Preprocessing?

Data preprocessing is the process of converting raw data into a
clean, organized, consistent, and suitable format for analysis
and machine learning.

It involves identifying and handling missing values, duplicate
records, inconsistent values, incorrect data types, and outliers.
It can also include encoding categorical variables, scaling
numerical features, and selecting relevant features.

**Example:** Converting missing age values into appropriate
imputed values and standardizing category labels before training
a machine learning model.



## 2. Why Is Data Preprocessing Required?

Data preprocessing is important because raw data may contain
errors, missing values, duplicates, inconsistent formats, and
irrelevant information.

Poor-quality data can lead to unreliable analysis and inaccurate
machine learning predictions.

The main benefits are:
- Improved data quality.
- More consistent and reliable results.
- Better model training and evaluation.
- Reduced errors caused by incorrect data.
- Prevention of data leakage.
- More reproducible machine learning workflows.


```python
raw_data = pd.DataFrame({
    "Name": ["Ravi", "Priya", "Ravi", "Arun"],
    "Age": [25, np.nan, 25, 200],
    "City": ["Hyderabad", "hyderabad", "Hyderabad", "Chennai"],
    "Salary": [30000, 45000, 30000, -5000]
})

print("Raw Dataset:")
display(raw_data)
```

```python
print("Dataset shape:", raw_data.shape)
print("\nMissing values:")
print(raw_data.isnull().sum())

print("\nDuplicate rows:", raw_data.duplicated().sum())

print("\nData types:")
print(raw_data.dtypes)
```


## 3. Data Quality

Data quality describes how suitable data is for its intended use.

### Completeness
Checks whether required values are present.

### Accuracy
Checks whether values correctly represent real-world information.

### Validity
Checks whether values follow defined rules, formats, and ranges.

### Consistency
Checks whether the same information is represented uniformly
across records and systems.

### Uniqueness
Checks whether duplicate records exist when each record should
be unique.

### Integrity
Ensures that relationships and constraints between data records
remain correct.

These dimensions help determine which cleaning operations are
necessary before using data in machine learning.


```python
print("City values before standardization:")
print(raw_data["City"].unique())

consistency_demo = raw_data.copy()
consistency_demo["City"] = (
    consistency_demo["City"]
    .str.strip()
    .str.title()
)

print("\nCity values after standardization:")
print(consistency_demo["City"].unique())
```

```python
missing_summary = pd.DataFrame({
    "Missing_Count": raw_data.isnull().sum(),
    "Missing_Percentage": (
        raw_data.isnull().mean() * 100
    ).round(2)
})

display(missing_summary)
```

```python
print("Age values outside the expected demo range (0–120):")
display(raw_data[
    raw_data["Age"].notna()
    & ~raw_data["Age"].between(0, 120)
])

print("\nNegative salaries:")
display(raw_data[raw_data["Salary"] < 0])
```


## 4. Data Integrity

Data integrity means maintaining the correctness and reliability
of data throughout collection, storage, transformation, and use.

Examples:
- Required fields must follow their defined rules.
- Customer IDs should be unique when required.
- Related records should reference valid IDs.
- Transformations should not unintentionally alter unrelated values.

Data integrity helps maintain trustworthy and reproducible results.



## 5. Data Leakage

Data leakage occurs when information unavailable at prediction
time influences model training or evaluation.

Examples:
- Using future information to predict a past outcome.
- Including the target variable among the input features.
- Imputing or scaling the entire dataset before splitting it.
- Using test-set information to select or tune a model.

To prevent leakage, split the data first. Fit preprocessing
transformations using training data only, then apply the fitted
transformations to validation and test data.



## 6. Training, Validation, and Test Data

**Training data:** Used to fit the machine learning model and
learn its parameters.

**Validation data:** Used to compare models, tune hyperparameters,
and make development decisions.

**Test data:** Used for final evaluation on data that was not
used for training or model selection.

A common split is 70% training, 15% validation, and 15% testing.
The appropriate split depends on dataset size, the task, and
whether the data has a time or group structure.


```python
from sklearn.model_selection import train_test_split

demo = pd.DataFrame({
    "feature": range(100),
    "target": [i * 2 for i in range(100)]
})

train_data, temporary_data = train_test_split(
    demo,
    test_size=0.30,
    random_state=42
)

validation_data, test_data = train_test_split(
    temporary_data,
    test_size=0.50,
    random_state=42
)

print("Training rows:", len(train_data))
print("Validation rows:", len(validation_data))
print("Test rows:", len(test_data))
```


## 7. Complete Data Preprocessing Workflow

The preprocessing workflow follows these stages:

1. Identify the dataset and understand its columns.
2. Inspect data types, missing values, duplicates, and unusual values.
3. Document the issues found during Exploratory Data Analysis.
4. Split the data into training, validation, and test sets.
5. Impute missing values using training data.
6. Encode categorical features and scale numerical features when needed.
7. Validate the transformed data.
8. Save the appropriate outputs and document the results.

The exact steps depend on the dataset and the machine learning task.


```python
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler

numerical_features = ["Age", "Salary"]
categorical_features = ["City"]

numerical_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer(transformers=[
    ("numerical", numerical_pipeline, numerical_features),
    ("categorical", categorical_pipeline, categorical_features)
])

print("Preprocessing pipeline created successfully!")
```

```python
pipeline_data = raw_data.copy()

# Remove exact duplicate rows for this demonstration.
pipeline_data = pipeline_data.drop_duplicates().copy()

# Apply a consistent category format.
pipeline_data["City"] = (
    pipeline_data["City"].str.strip().str.title()
)

# Apply the demo's documented validity rule.
pipeline_data.loc[
    ~pipeline_data["Age"].between(0, 120),
    "Age"
] = np.nan

pipeline_data.loc[
    pipeline_data["Salary"] < 0,
    "Salary"
] = np.nan

# Demonstration only: use the remaining data as a training sample.
X_demo = pipeline_data[numerical_features + categorical_features]

transformed = preprocessor.fit_transform(X_demo)

print("Original feature shape:", X_demo.shape)
print("Transformed feature shape:", transformed.shape)
print("Pipeline executed successfully!")
```

```python
print("Rows after duplicate removal:", len(pipeline_data))
print("Remaining duplicate rows:", pipeline_data.duplicated().sum())

print("\nMissing values after applying demo cleaning rules:")
print(pipeline_data.isnull().sum())

print("\nFinal data types:")
print(pipeline_data.dtypes)

assert pipeline_data.duplicated().sum() == 0

print("\nValidation checks completed.")
```
